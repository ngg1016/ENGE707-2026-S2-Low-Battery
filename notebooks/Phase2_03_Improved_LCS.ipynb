# Phase II: Improved LCS System

## 1. Load Preprocessed Datasets

```python
import pandas as pd

train_data = pd.read_csv(
    "../data/feature_engineered/telecom_train_preprocessed_all_features.csv"
)

test_data = pd.read_csv(
    "../data/feature_engineered/telecom_test_preprocessed_all_features.csv"
)

print(train_data.shape)
print(test_data.shape)
```

```python
X_train = train_data.drop(
    columns=["Churn Target"]
)

y_train = train_data["Churn Target"]

X_test = test_data.drop(
    columns=["Churn Target"]
)

y_test = test_data["Churn Target"]

print(X_train.shape)
print(X_test.shape)
```

## 2. Original eLCS on Preprocessed Data

```python
from skeLCS import eLCS
```

```python
baseline_preprocessed_lcs = eLCS()

baseline_preprocessed_lcs.fit(
    X_train.to_numpy(),
    y_train.to_numpy()
)

print("Training completed successfully")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix
)

preprocessed_lcs_predictions = baseline_preprocessed_lcs.predict(
    X_test.to_numpy()
)

preprocessed_lcs_accuracy = accuracy_score(
    y_test,
    preprocessed_lcs_predictions
)

preprocessed_lcs_precision = precision_score(
    y_test,
    preprocessed_lcs_predictions,
    zero_division=0
)

preprocessed_lcs_recall = recall_score(
    y_test,
    preprocessed_lcs_predictions,
    zero_division=0
)

preprocessed_lcs_f1 = f1_score(
    y_test,
    preprocessed_lcs_predictions,
    zero_division=0
)

preprocessed_lcs_balanced_accuracy = balanced_accuracy_score(
    y_test,
    preprocessed_lcs_predictions
)

print("Original eLCS on Preprocessed Data")
print("-----------------------------------")
print("Accuracy:", round(preprocessed_lcs_accuracy, 4))
print("Precision:", round(preprocessed_lcs_precision, 4))
print("Recall:", round(preprocessed_lcs_recall, 4))
print("F1 Score:", round(preprocessed_lcs_f1, 4))
print(
    "Balanced Accuracy:",
    round(preprocessed_lcs_balanced_accuracy, 4)
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        preprocessed_lcs_predictions
    )
)


```

```python
from pathlib import Path
import pandas as pd

preprocessed_lcs_directory = Path(
    "../results/preprocessed_lcs"
)

preprocessed_lcs_directory.mkdir(
    parents=True,
    exist_ok=True
)

preprocessed_lcs_metrics = pd.DataFrame({
    "Model": ["Original eLCS on Preprocessed Data"],
    "Accuracy": [preprocessed_lcs_accuracy],
    "Precision": [preprocessed_lcs_precision],
    "Recall": [preprocessed_lcs_recall],
    "F1 Score": [preprocessed_lcs_f1],
    "Balanced Accuracy": [
        preprocessed_lcs_balanced_accuracy
    ]
})

preprocessed_lcs_metrics.to_csv(
    preprocessed_lcs_directory
    / "preprocessed_lcs_metrics.csv",
    index=False
)

preprocessed_lcs_predictions_df = pd.DataFrame({
    "actual_target": y_test.to_numpy(),
    "predicted_target": preprocessed_lcs_predictions
})

preprocessed_lcs_predictions_df.to_csv(
    preprocessed_lcs_directory
    / "preprocessed_lcs_predictions.csv",
    index=False
)

print("Preprocessed eLCS results exported.")
```

## 3. Improved eLCS Configuration

```python
improved_lcs = eLCS(
    learning_iterations=15000,
    N=2000,
    theta_GA=50,
    random_state=42
)

print(improved_lcs.get_params())
```

## 4. Improved eLCS Training

```python
import time

training_start_time = time.time()

improved_lcs.fit(
    X_train.to_numpy(),
    y_train.to_numpy()
)

training_end_time = time.time()

improved_training_time = (
    training_end_time - training_start_time
)

print("Improved eLCS training completed.")

print(
    "Training time:",
    round(improved_training_time, 2),
    "seconds"
)
```

## 5. Improved eLCS Evaluation

```python
improved_predictions = improved_lcs.predict(
    X_test.to_numpy()
)

print("Predictions generated.")
print("Number of predictions:", len(improved_predictions))
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    classification_report
)

improved_accuracy = accuracy_score(
    y_test,
    improved_predictions
)

improved_precision = precision_score(
    y_test,
    improved_predictions,
    zero_division=0
)

improved_recall = recall_score(
    y_test,
    improved_predictions,
    zero_division=0
)

improved_f1 = f1_score(
    y_test,
    improved_predictions,
    zero_division=0
)

improved_balanced_accuracy = (
    balanced_accuracy_score(
        y_test,
        improved_predictions
    )
)

print("Improved eLCS Results")
print("----------------------")

print("Accuracy:", round(improved_accuracy, 4))
print("Precision:", round(improved_precision, 4))
print("Recall:", round(improved_recall, 4))
print("F1 Score:", round(improved_f1, 4))
print(
    "Balanced Accuracy:",
    round(improved_balanced_accuracy, 4)
)

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        improved_predictions
    )
)

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        improved_predictions,
        zero_division=0
    )
)
```
