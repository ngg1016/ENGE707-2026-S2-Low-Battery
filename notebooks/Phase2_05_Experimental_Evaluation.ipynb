## 6. Statistical Comparison

```python
improved_predictions_data = pd.read_csv(
    "../results/improved_lcs/improved_lcs_predictions.csv"
)

conventional_predictions_data = pd.read_csv(
    "../results/model_comparison/"
    "conventional_model_predictions.csv"
)

actual_target = conventional_predictions_data[
    "actual_target"
].to_numpy()

improved_predictions = improved_predictions_data[
    "predicted_target"
].to_numpy()

random_forest_predictions = conventional_predictions_data[
    "random_forest_prediction"
].to_numpy()

print("Actual targets:", len(actual_target))
print("Improved eLCS predictions:", len(improved_predictions))
print("Random Forest predictions:", len(random_forest_predictions))
```

```python
from scipy.stats import binomtest

improved_correct = (
    improved_predictions == actual_target
)

random_forest_correct = (
    random_forest_predictions == actual_target
)

improved_only_correct = (
    improved_correct
    & ~random_forest_correct
).sum()

random_forest_only_correct = (
    random_forest_correct
    & ~improved_correct
).sum()

discordant_total = (
    improved_only_correct
    + random_forest_only_correct
)

print(
    "Improved eLCS only correct:",
    improved_only_correct
)

print(
    "Random Forest only correct:",
    random_forest_only_correct
)

print(
    "Total discordant predictions:",
    discordant_total
)

if discordant_total > 0:
    mcnemar_result = binomtest(
        k=min(
            improved_only_correct,
            random_forest_only_correct
        ),
        n=discordant_total,
        p=0.5,
        alternative="two-sided"
    )

    print(
        "Exact McNemar p-value:",
        mcnemar_result.pvalue
    )
else:
    print(
        "McNemar test could not be performed because "
        "there were no discordant predictions."
    )
```

```python
statistical_test_results = pd.DataFrame({
    "Model A": ["Improved eLCS"],
    "Model B": ["Random Forest"],
    "Model A only correct": [improved_only_correct],
    "Model B only correct": [random_forest_only_correct],
    "Discordant predictions": [discordant_total],
    "Exact McNemar p-value": [
        mcnemar_result.pvalue if discordant_total > 0 else None
    ]
})

statistical_test_results.to_csv(
    evaluation_dir / "mcnemar_test_results.csv",
    index=False
)

display(statistical_test_results)
```
